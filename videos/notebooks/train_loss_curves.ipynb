# Training Loss Curves

Plot validation loss + accuracy curves from the per-subject training logs (`logs_s{i}.csv`) emitted by `videos/scripts/train_cv.py`, plus the general-split log (`logs_general.csv`) emitted by `videos/scripts/train.py`. Reports the best per-fold accuracy and the std across folds.

## Configuration

```python
PER_SUBJECT_LOGS_GLOB = "logs_s*.csv"
GENERAL_LOG = "logs_general.csv"
```

## Setup

```python
import pandas as pd
import matplotlib.pyplot as plt
import os, glob
import numpy as np
plt.style.use('seaborn-v0_8-pastel')
```

## Inspect a single subject's log

```python
handle = pd.read_csv("logs_s0.csv"  # single-subject preview)
```

## Plot per-subject curves

```python
#Change the plt style

all_files = glob.glob(PER_SUBJECT_LOGS_GLOB)
fig,ax = plt.subplots(1,2,figsize=(10,5))

v = []
for file in all_files:
    handle = pd.read_csv(file)
    name = file.split('.')[0].split('_')[-1][-1]
    if not name.isdigit():
        continue
    if name == '0':
        name = '5'

    best_value = handle['eval_accuracy'].argmax()
    best_epoch = handle['epoch'][best_value]

    v.append(handle['eval_accuracy'][best_value])


    ax[0].plot(handle['epoch'],handle['eval_loss'],label='Subject '+name)
    ax[1].plot(handle['epoch'],handle['eval_accuracy'],label='Subject '+name)
    ax[1].text(best_epoch,handle['eval_accuracy'][best_value],str(round(handle['eval_accuracy'][best_value],2)),fontsize=8)
    ax[0].set_title('Validation Loss')
    ax[0].legend()
    ax[0].grid()
    ax[0].set_xlabel('Epoch')
    ax[0].set_ylabel('Loss')
    ax[1].set_title('Validation Accuracy')
    ax[1].legend()
    ax[1].set_xlabel('Epoch')
    ax[1].set_ylabel('Accuracy')
    ax[1].grid()
fig.suptitle('Model Evaluation (LOSO)')
plt.show()

np.mean(v)

```

## Best-accuracy std across folds

```python
import numpy as np

np.std([0.34,0.38,0.39,0.44,0.26])
```

## General-split curves

```python
handle = pd.read_csv(GENERAL_LOG)
fig,ax = plt.subplots(1,2,figsize=(10,5))
ax[0].plot(handle['epoch'],handle['eval_loss'])
ax[1].plot(handle['epoch'],handle['eval_accuracy'])
ax[0].set_title('Validation Loss')
ax[0].grid()
ax[0].set_xlabel('Epoch')
ax[0].set_ylabel('Loss')
ax[1].set_title('Validation Accuracy')
ax[1].grid()
ax[1].set_xlabel('Epoch')
ax[1].set_ylabel('Accuracy')
fig.suptitle('Model Evaluation')
plt.show()
```
