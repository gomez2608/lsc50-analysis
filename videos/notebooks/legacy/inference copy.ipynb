```python
import torch
import os, glob
from transformers import VivitForVideoClassification
from datasets import load_dataset
```

```python
ds = load_dataset("dataset",split="test")
```

```python
model = VivitForVideoClassification.from_pretrained("vivit-finetuned")
model.cuda()
```

```python

with torch.no_grad():
    outputs = model(ds["test"]["pixel_values"].cuda())
    logits = outputs.logits
```

```python
from sklearn.metrics import classification_report, confusion_matrix

preds = []
for pred in range(logits.shape[0]):
    vals = logits[pred]
    preds.append(torch.argmax(vals).numpy())
```
